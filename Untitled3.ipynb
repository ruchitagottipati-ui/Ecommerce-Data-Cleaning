# SQL Analysis: Amazon E-Commerce Sales Dataset

**Goal:** Answer the same kinds of business questions explored in Week 2 (Pandas EDA), but written in SQL — the query language every data role expects. All 10 queries run against a SQLite database built from the same cleaned dataset (128,975 rows, 0 nulls, 0 duplicates) from Week 1.

Results are cross-checked against Week 2's Pandas findings throughout, to confirm accuracy.

## Setup: Load cleaned data and create SQL table

Re-applying the Week 1 cleaning pipeline to rebuild the same clean dataset, then loading it into a SQLite table called `orders`.


```python
import pandas as pd
import sqlite3

# Load original dataset
sheet_id = "1VD2q40muLWtXPoItm2OKlsgKYTikLU3m"
url = f"https://docs.google.com/spreadsheets/d/{sheet_id}/export?format=csv"
df = pd.read_csv(url, low_memory=False)

# Apply Week 1 cleaning steps
df = df.drop(columns=['Unnamed: 22'])
df['fulfilled-by'] = df['fulfilled-by'].fillna('Amazon')
df['promotion-ids'] = df['promotion-ids'].fillna('No Promotion')
df['Courier Status'] = df['Courier Status'].fillna('Not Shipped')

cancelled_mask = (df['Amount'].isnull()) & (df['Status'] == 'Cancelled')
df.loc[cancelled_mask, 'Amount'] = 0
df.loc[cancelled_mask, 'currency'] = 'N/A'
df['Amount'] = df['Amount'].fillna(-1)
df['currency'] = df['currency'].fillna('Unknown')

location_cols = ['ship-city', 'ship-state', 'ship-country']
df[location_cols] = df[location_cols].fillna('Unknown')
df['ship-postal-code'] = df['ship-postal-code'].fillna(0)

df = df.drop_duplicates()

df['Date'] = pd.to_datetime(df['Date'], format='mixed', dayfirst=False)
df['ship-postal-code'] = df['ship-postal-code'].astype('Int64')

print("Nulls:", df.isnull().sum().sum(), "| Rows:", len(df))
```

```python
conn = sqlite3.connect('ecommerce.db')
df.to_sql('orders', conn, if_exists='replace', index=False

print("Table 'orders' created successfully")
```

## Query 1: How many orders fall under each Status?

Simple aggregation using `GROUP BY` — a good starting point, and directly comparable against Week 2's `value_counts()` output to confirm the SQL matches Pandas.

**Verified against Pandas:** counts match exactly (Shipped: 77,804 | Delivered: 28,769 | Cancelled: 18,332).

```python
query1 = """
SELECT Status, COUNT(*) AS order_count
FROM orders
GROUP BY Status
ORDER BY order_count DESC;
"""

result1 = pd.read_sql_query(query1, conn)
print("Q1: How many orders fall under each Status?")
result1
```

## Query 2: Total and average order value by Category

Combines multiple aggregate functions (`COUNT`, `SUM`, `AVG`) in one query. Filters out cancelled/unknown-amount orders (`Amount > 0`) — same logic used for the Week 2 histogram — so the averages reflect real completed purchases only.

**Insight:** `Set` leads in both order count *and* total revenue, but `Western Dress` has a notably higher average order value despite fewer orders — a smaller but higher-value category.

```python
query2 = """
SELECT Category,
       COUNT(*) AS order_count,
       ROUND(SUM(Amount), 2) AS total_amount,
       ROUND(AVG(Amount), 2) AS avg_amount
FROM orders
WHERE Amount > 0
GROUP BY Category
ORDER BY total_amount DESC;
"""

result2 = pd.read_sql_query(query2, conn)
print("Q2: What's the total and average order value per Category?")
result2
```

## Query 3: JOIN — group granular statuses into broader categories

Builds a small lookup table (`status_lookup`) mapping each of the 13 specific Status values into 5 broader groups (Completed, Cancelled, Returned, Issue, Pending), then JOINs it to `orders`.

**Verified against Pandas:** Cancelled count (18,332) matches Week 2 exactly.

**Insight:** ~83% of orders complete successfully, 14% are cancelled — consistent with Week 2 findings.

```python
# Step 1: Create a small lookup table mapping each Status to a broader category
status_mapping = pd.DataFrame({
    'Status': ['Shipped', 'Shipped - Delivered to Buyer', 'Shipped - Picked Up',
               'Shipped - Out for Delivery', 'Cancelled', 'Shipped - Returned to Seller',
               'Shipped - Returning to Seller', 'Shipped - Rejected by Buyer',
               'Shipped - Lost in Transit', 'Shipped - Damaged', 'Pending',
               'Pending - Waiting for Pick Up', 'Shipping'],
    'status_group': ['Completed', 'Completed', 'Completed', 'Completed', 'Cancelled',
                      'Returned', 'Returned', 'Returned', 'Issue', 'Issue',
                      'Pending', 'Pending', 'Pending']
})

status_mapping.to_sql('status_lookup', conn, if_exists='replace', index=False)

# Step 2: JOIN orders with this lookup table
query3 = """
SELECT s.status_group, COUNT(*) AS order_count
FROM orders o
JOIN status_lookup s ON o.Status = s.Status
GROUP BY s.status_group
ORDER BY order_count DESC;
"""

result3 = pd.read_sql_query(query3, conn)
print("Q3: How many orders fall into each broader status group (Completed/Cancelled/Returned/Issue/Pending)?")
result3
```

## Query 4: Monthly order and revenue trend

Uses `strftime('%Y-%m', Date)` to collapse daily dates into monthly buckets, plus a `CASE WHEN` to safely sum only valid (non-cancelled) revenue.

**Insight:** April was the strongest month overall (49,067 orders, ₹2.88 crore revenue), with a declining trend into May and June — complements Week 2's daily chart, which showed a spike in early May but not this bigger monthly decline.

```python
query4 = """
SELECT strftime('%Y-%m', Date) AS order_month,
       COUNT(*) AS order_count,
       ROUND(SUM(CASE WHEN Amount > 0 THEN Amount ELSE 0 END), 2) AS total_revenue
FROM orders
GROUP BY order_month
ORDER BY order_month;
"""

result4 = pd.read_sql_query(query4, conn)
print("Q4: How do order count and revenue trend month over month?")
result4
```

## Query 5: Categories priced above the overall average (subquery)

Uses a subquery — `(SELECT AVG(Amount) FROM orders WHERE Amount > 0)` — to calculate one benchmark number, then filters categories against it using `HAVING`.

**Insight:** Only 4 of 8 categories (`Set`, `Saree`, `Western Dress`, `Ethnic Dress`) exceed the overall average order value. Notably, `kurta` — the #2 category by volume — sells below average, showing high volume doesn't always mean high value.

```python
query5 = """
SELECT Category, ROUND(AVG(Amount), 2) AS avg_category_amount
FROM orders
WHERE Amount > 0
GROUP BY Category
HAVING AVG(Amount) > (
    SELECT AVG(Amount) FROM orders WHERE Amount > 0
)
ORDER BY avg_category_amount DESC;
"""

result5 = pd.read_sql_query(query5, conn)
print("Q5: Which categories have an average order value above the overall average?")
result5
```

## Query 6: Multiple JOINs — price tier vs. order outcome

Adds a second lookup table (`category_tier`, tagging categories as Premium/Standard based on Query 5's result) and JOINs both lookup tables to `orders` simultaneously.

**Insight:** Cancellation rates are nearly identical between Premium (~14.3%) and Standard (~13.9%) tiers — price tier doesn't meaningfully predict cancellation behavior, suggesting the cause lies elsewhere (e.g. checkout experience) rather than product pricing.

```python
# Create a second small lookup: flag high-value vs standard categories
category_tier = pd.DataFrame({
    'Category': ['Set', 'kurta', 'Western Dress', 'Top', 'Ethnic Dress', 'Blouse', 'Bottom', 'Saree', 'Dupatta'],
    'price_tier': ['Premium', 'Standard', 'Premium', 'Standard', 'Premium', 'Standard', 'Standard', 'Premium', 'Standard']
})
category_tier.to_sql('category_tier', conn, if_exists='replace', index=False)

query6 = """
SELECT c.price_tier, s.status_group, COUNT(*) AS order_count
FROM orders o
JOIN category_tier c ON o.Category = c.Category
JOIN status_lookup s ON o.Status = s.Status
GROUP BY c.price_tier, s.status_group
ORDER BY c.price_tier, order_count DESC;
"""

result6 = pd.read_sql_query(query6, conn)
print("Q6: Do Premium-tier categories have different cancellation/completion patterns than Standard-tier?")
result6
```

## Query 7: Top 5 cities by revenue

Straightforward `ORDER BY` + `LIMIT` to rank cities, excluding the "Unknown" placeholder from Week 1 cleaning.

**Insight:** Bengaluru leads decisively (₹68.5 lakh revenue), nearly 40% ahead of #2 Hyderabad.

```python
query7 = """
SELECT [ship-city] AS city,
       COUNT(*) AS order_count,
       ROUND(SUM(Amount), 2) AS total_revenue
FROM orders
WHERE Amount > 0 AND [ship-city] != 'Unknown'
GROUP BY [ship-city]
ORDER BY total_revenue DESC
LIMIT 5;
"""

result7 = pd.read_sql_query(query7, conn)
print("Q7: Which 5 cities generate the most revenue?")
result7
```

## Query 8: Orders priced above their own category's average (CTE)

Originally written as a correlated subquery (recalculating each category's average once per row — extremely slow at ~129K rows). Rewritten using a **CTE** (`WITH category_avg AS (...)`) to precompute each category's average once, then JOIN it back — same logical result, dramatically faster.

**Insight:** Raw counts are highest for `kurta` and `Set`, but that's expected given their much larger total order volumes — a reminder to always check counts relative to their total base, not in isolation.

```python
query8_fast = """
WITH category_avg AS (
    SELECT Category, AVG(Amount) AS avg_amount
    FROM orders
    WHERE Amount > 0
    GROUP BY Category
)
SELECT o.Category, COUNT(*) AS above_avg_orders
FROM orders o
JOIN category_avg c ON o.Category = c.Category
WHERE o.Amount > c.avg_amount AND o.Amount > 0
GROUP BY o.Category
ORDER BY above_avg_orders DESC;
"""

result8 = pd.read_sql_query(query8_fast, conn)
print("Q8: How many orders in each category are priced above that category's own average?")
result8
```

## Query 9: B2B vs. regular customer orders

Groups by the boolean `B2B` column to compare business vs. retail customer behavior.

**Insight:** B2B orders (under 1% of total) are only marginally higher in value (₹701 vs ₹661) and quantity (1.00 vs 0.96) than regular orders — no dramatic bulk-buying pattern, suggesting B2B accounts here are small businesses ordering individually rather than wholesale buyers.

```python
query9 = """
SELECT B2B,
       COUNT(*) AS order_count,
       ROUND(AVG(Amount), 2) AS avg_order_value,
       ROUND(AVG(Qty), 2) AS avg_quantity
FROM orders
WHERE Amount > 0
GROUP BY B2B;
"""

result9 = pd.read_sql_query(query9, conn)
print("Q9: Do B2B (business) orders differ from regular customer orders in value and quantity?")
result9
```

## Query 10: Cancellation rate by category

Combines a JOIN (to get status groupings), conditional aggregation (`SUM(CASE WHEN...)`), and a calculated percentage — bringing together every technique used across this analysis.

**Insight:** Cancellation rates are consistent across categories (12-15%), with no category standing out as a major outlier. This reinforces Query 6's finding: cancellations appear to be a systemic issue (checkout/payment/delivery-related) rather than tied to specific products.

```python
query10 = """
SELECT o.Category,
       COUNT(*) AS total_orders,
       SUM(CASE WHEN s.status_group = 'Cancelled' THEN 1 ELSE 0 END) AS cancelled_orders,
       ROUND(100.0 * SUM(CASE WHEN s.status_group = 'Cancelled' THEN 1 ELSE 0 END) / COUNT(*), 2) AS cancellation_rate_pct
FROM orders o
JOIN status_lookup s ON o.Status = s.Status
GROUP BY o.Category
ORDER BY cancellation_rate_pct DESC;
"""

result10 = pd.read_sql_query(query10, conn)
print("Q10: Which product categories have the highest cancellation rate?")
result10
```

## Summary

- Built a SQLite database from the Week 1 cleaned dataset (128,975 rows, 0 nulls, 0 duplicates).
- Wrote 10 analytical queries covering `GROUP BY`, aggregations (`COUNT`, `SUM`, `AVG`), multiple `JOIN`s across 3 tables, subqueries (both scalar and CTE-based), and conditional logic (`CASE WHEN`).
- Cross-verified key results (Status counts, Cancelled totals) against Week 2's Pandas findings — all matched.
- Key findings: cancellation rate (~14%) is consistent across categories and price tiers, pointing to a systemic rather than product-specific cause; `Set` and `kurta` dominate volume but `kurta` underperforms on average order value; Bengaluru is the clear revenue leader; B2B orders behave similarly to retail orders despite the B2B label.

```python
%%writefile queries.sql
-- Q1: How many orders fall under each Status?
SELECT Status, COUNT(*) AS order_count
FROM orders
GROUP BY Status
ORDER BY order_count DESC;

-- Q2: Total and average order value by Category
SELECT Category,
       COUNT(*) AS order_count,
       ROUND(SUM(Amount), 2) AS total_amount,
       ROUND(AVG(Amount), 2) AS avg_amount
FROM orders
WHERE Amount > 0
GROUP BY Category
ORDER BY total_amount DESC;

-- Q3: Broader status groups (requires status_lookup table)
SELECT s.status_group, COUNT(*) AS order_count
FROM orders o
JOIN status_lookup s ON o.Status = s.Status
GROUP BY s.status_group
ORDER BY order_count DESC;

-- Q4: Monthly order and revenue trend
SELECT strftime('%Y-%m', Date) AS order_month,
       COUNT(*) AS order_count,
       ROUND(SUM(CASE WHEN Amount > 0 THEN Amount ELSE 0 END), 2) AS total_revenue
FROM orders
GROUP BY order_month
ORDER BY order_month;

-- Q5: Categories priced above the overall average (subquery)
SELECT Category, ROUND(AVG(Amount), 2) AS avg_category_amount
FROM orders
WHERE Amount > 0
GROUP BY Category
HAVING AVG(Amount) > (
    SELECT AVG(Amount) FROM orders WHERE Amount > 0
)
ORDER BY avg_category_amount DESC;

-- Q6: Price tier vs order outcome (requires category_tier table)
SELECT c.price_tier, s.status_group, COUNT(*) AS order_count
FROM orders o
JOIN category_tier c ON o.Category = c.Category
JOIN status_lookup s ON o.Status = s.Status
GROUP BY c.price_tier, s.status_group
ORDER BY c.price_tier, order_count DESC;

-- Q7: Top 5 cities by revenue
SELECT [ship-city] AS city,
       COUNT(*) AS order_count,
       ROUND(SUM(Amount), 2) AS total_revenue
FROM orders
WHERE Amount > 0 AND [ship-city] != 'Unknown'
GROUP BY [ship-city]
ORDER BY total_revenue DESC
LIMIT 5;

-- Q8: Orders priced above their own category's average (CTE)
WITH category_avg AS (
    SELECT Category, AVG(Amount) AS avg_amount
    FROM orders
    WHERE Amount > 0
    GROUP BY Category
)
SELECT o.Category, COUNT(*) AS above_avg_orders
FROM orders o
JOIN category_avg c ON o.Category = c.Category
WHERE o.Amount > c.avg_amount AND o.Amount > 0
GROUP BY o.Category
ORDER BY above_avg_orders DESC;

-- Q9: B2B vs regular customer orders
SELECT B2B,
       COUNT(*) AS order_count,
       ROUND(AVG(Amount), 2) AS avg_order_value,
       ROUND(AVG(Qty), 2) AS avg_quantity
FROM orders
WHERE Amount > 0
GROUP BY B2B;

-- Q10: Cancellation rate by category
SELECT o.Category,
       COUNT(*) AS total_orders,
       SUM(CASE WHEN s.status_group = 'Cancelled' THEN 1 ELSE 0 END) AS cancelled_orders,
       ROUND(100.0 * SUM(CASE WHEN s.status_group = 'Cancelled' THEN 1 ELSE 0 END) / COUNT(*), 2) AS cancellation_rate_pct
FROM orders o
JOIN status_lookup s ON o.Status = s.Status
GROUP BY o.Category
ORDER BY cancellation_rate_pct DESC;


```
