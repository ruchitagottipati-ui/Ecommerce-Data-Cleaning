# Data Cleaning: Amazon E-Commerce Sales Dataset

**Dataset source:** Amazon Sale Report — real e-commerce order data (128,975 rows, 24 columns), originally sourced from Kaggle ("Unlock Profits with E-Commerce Sales Data").

**Goal:** Clean this real-world messy dataset using Pandas — handling missing values, removing duplicates, and fixing data types — with every decision backed by evidence, not guesswork.

```python
import pandas as pd

sheet_id = "1VD2q40muLWtXPoItm2OKlsgKYTikLU3m"
url = f"https://docs.google.com/spreadsheets/d/{sheet_id}/export?format=csv"
df = pd.read_csv(url, low_memory=False)

df.head()
```

## 1. Initial Inspection

```python
print("Shape:", df.shape)
df.info()
```

## 2. BEFORE Stats

Capturing baseline numbers before any cleaning, so we can prove the cleaning actually improved the dataset.

```python
before_rows = len(df)
before_nulls = df.isnull().sum().sum()
before_duplicates = df.duplicated().sum()

print("BEFORE — rows:", before_rows)
print("BEFORE — total nulls:", before_nulls)
print("BEFORE — duplicates:", before_duplicates)

df.isnull().sum().sort_values(ascending=False)
```

## 3. Handling Missing Values

Every decision below is backed by checking the actual data first — not guessed.

### `Unnamed: 22` — drop the column

This column has no real header name and is 38% missing. Checking `value_counts()` showed it contains **only** `0.0` or blank — zero variation across all 128,975 rows. A column that never changes carries no analytical information, so we drop it entirely rather than guessing what to fill it with.

```python
print(df['Unnamed: 22'].value_counts())
df = df.drop(columns=['Unnamed: 22'])
```

### `fulfilled-by` — fill with `'Amazon'`

70% missing at first glance looks alarming — but a crosstab against `Fulfilment` showed a perfect 100% pattern: `fulfilled-by` is missing **exactly** when `Fulfilment == 'Amazon'`, and always filled in when `Fulfilment == 'Merchant'`. This isn't a data error — it's a field that only applies to merchant-fulfilled orders, since Amazon doesn't need to name a third-party fulfillment partner for its own orders.

```python
print(pd.crosstab(df['fulfilled-by'].isnull(), df['Fulfilment']))
df['fulfilled-by'] = df['fulfilled-by'].fillna('Amazon')
```

### `promotion-ids` — fill with `'No Promotion'`

This column logs which promotional offers applied to an order — a highly variable, order-specific text field (some rows list a dozen+ promo codes at once). A missing value here almost certainly means no promotion applied to that order (bought at standard price), not a data collection error.

```python
df['promotion-ids'] = df['promotion-ids'].fillna('No Promotion')
```

### `Courier Status` — fill with `'Not Shipped'`

A crosstab against `Status` showed 99.8% of missing `Courier Status` values belong to `Cancelled` orders. This makes sense: a cancelled order was never handed to a courier, so there's nothing to record. Filling with `'Not Shipped'` reflects this reality rather than a vague `'Unknown'`.

```python
print(pd.crosstab(df['Courier Status'].isnull(), df['Status']))
df['Courier Status'] = df['Courier Status'].fillna('Not Shipped')
```

### `Amount` / `currency` — split strategy based on evidence

Checked `Status` for rows missing `Amount`: 97% were `Cancelled` orders (no completed sale = no amount to record, which is expected). The remaining ~3% were on non-cancelled orders (Shipped, Pending, etc.) — genuine data gaps rather than expected absence. These two cases are handled differently: cancelled orders get `0` (no sale occurred), while true gaps get `-1` as a clear "unknown" flag, since a real Amount can never be negative.

```python
print(df[df['Amount'].isnull()]['Status'].value_counts())

cancelled_mask = (df['Amount'].isnull()) & (df['Status'] == 'Cancelled')
df.loc[cancelled_mask, 'Amount'] = 0
df.loc[cancelled_mask, 'currency'] = 'N/A'
df['Amount'] = df['Amount'].fillna(-1)
df['currency'] = df['currency'].fillna('Unknown')
```

### `ship-city` / `ship-state` / `ship-postal-code` / `ship-country` — fill with `'Unknown'`

Only 33 rows (0.026% of the dataset) are missing these fields, with no clear pattern tying them to a specific `Status` — likely random data-entry gaps. Given the negligible count and lack of pattern, filling with `'Unknown'` preserves these rows for analysis on other columns rather than discarding them.

```python
location_cols = ['ship-city', 'ship-state', 'ship-country']
df[location_cols] = df[location_cols].fillna('Unknown')
df['ship-postal-code'] = df['ship-postal-code'].fillna(0)

print("Remaining nulls:", df.isnull().sum().sum())
```

## 4. Removing Duplicate Rows

Checked for exact duplicate rows (every column matching another row) — unlike missing values, these rarely need judgment, since an exact duplicate is almost always a data export/entry error rather than two genuinely different events.

```python
duplicates_found = df.duplicated().sum()
print("Duplicate rows found:", duplicates_found)

df = df.drop_duplicates()
print("Rows after removing duplicates:", len(df))
```

## 5. Fixing Data Types

- **`Date`** was stored as text, and inspection revealed a mix of two formats (plain strings like `'04-30-22'` and already-converted datetime objects) — likely from the file passing through Excel inconsistently. Converted using `format='mixed'` with `dayfirst=False`, matching the confirmed MM-DD-YY order in the string dates.
- **`ship-postal-code`** was stored as a float (e.g. `400081.0`) despite being an identifier, not a number to calculate with. Converted to pandas' nullable `Int64` type to remove the misleading decimal.

```python
df['Date'] = pd.to_datetime(df['Date'], format='mixed', dayfirst=False)
df['ship-postal-code'] = df['ship-postal-code'].astype('Int64')

df.dtypes
```

## 6. AFTER Stats — Before vs After Comparison

```python
after_rows = len(df)
after_nulls = df.isnull().sum().sum()
after_duplicates = df.duplicated().sum()

comparison = pd.DataFrame({
    'Metric': ['Row count', 'Total missing cells', 'Duplicate rows'],
    'Before': [before_rows, before_nulls, before_duplicates],
    'After': [after_rows, after_nulls, after_duplicates]
})
comparison
```

## Summary

- Started with **128,975 rows** and **210,495 missing cells** spread across 10 columns.
- Dropped 1 column (`Unnamed: 22`) that carried zero information — confirmed via `value_counts()` before deciding.
- Filled every other missing value using evidence gathered from the data itself (crosstabs, value counts, status checks) rather than guessing — e.g. `fulfilled-by` missing = Amazon-fulfilled, `Amount` missing = mostly cancelled orders.
- Removed exact duplicate rows (0 found, confirmed and documented).
- Fixed `Date` (text → datetime, handling a mixed-format issue) and `ship-postal-code` (float → nullable Int64).
- Result: **0 nulls, 0 duplicates**, correct data types — ready for analysis.

## Descriptive Statistics & Visualizations

Now that the data is clean, let's explore it to find patterns worth reporting.

### Chart 1: How are orders distributed across Status?

```python
import matplotlib.pyplot as plt

status_counts = df['Status'].value_counts()

plt.figure(figsize=(10, 6))
status_counts.plot(kind='bar', color='steelblue')
plt.title('Order Count by Status')
plt.xlabel('Status')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Chart 2: Which product categories sell the most?

```python
category_counts = df['Category'].value_counts()

plt.figure(figsize=(10, 6))
category_counts.plot(kind='bar', color='coral')
plt.title('Order Count by Product Category')
plt.xlabel('Category')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

**Insight:** Two categories — `Set` (50,284 orders) and `kurta` (49,877 orders) — dominate the business, together accounting for roughly 78% of all orders. `Western Dress` and `Top` form a distant second tier (15,500 and 10,622 respectively), while the remaining five categories (Ethnic Dress, Blouse, Bottom, Saree, Dupatta) are nearly negligible, together making up less than 2% of total orders. This suggests the business is heavily concentrated in just two product lines — a risk if demand shifts, but also a clear signal of where current inventory and marketing focus should stay.

### Chart 3: How is order value (Amount) distributed?


```python
# Exclude cancelled orders (Amount = 0) and unknown-amount flags (-1) for a meaningful distribution
valid_amounts = df[df['Amount'] > 0]['Amount']

plt.figure(figsize=(10, 6))
valid_amounts.plot(kind='hist', bins=50, color='mediumseagreen', edgecolor='black')
plt.title('Distribution of Order Amount (Completed Orders)')
plt.xlabel('Order Amount (INR)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

**Insight:** Order amounts are heavily concentrated in the ₹300–₹900 range, with a peak around ₹500 — matching closely with the median (₹583) calculated earlier. The distribution is right-skewed: a long tail of orders stretches out toward ₹2,000–₹5,000+, but these high-value orders are rare compared to the dense cluster of typical, lower-cost purchases. This confirms the business's core revenue comes from frequent, moderately-priced orders rather than occasional big-ticket sales — useful to know since strategies to grow revenue (e.g. upselling, bundling) might focus on nudging typical orders upward, rather than chasing rare high spenders.

### Chart 4: How do daily order counts trend over time?

```python
daily_orders = df.groupby(df['Date'].dt.date).size()

plt.figure(figsize=(12, 6))
daily_orders.plot(kind='line', color='darkorange')
plt.title('Daily Order Count Over Time (Apr - Jun 2022)')
plt.xlabel('Date')
plt.ylabel('Number of Orders')
plt.tight_layout()
plt.show()
```

**Insight:** Order volume shows a clear structural shift around early May: from late March through the end of April, daily orders climb and fluctuate around 500-700 per day, peaking sharply near 700-750 orders around May 1st. After that peak, orders drop noticeably and settle into a lower, more stable pattern (roughly 350-450 per day) from mid-May through late June. This suggests either a promotional event or seasonal demand spike drove the late-April/early-May surge, followed by a return to baseline — worth investigating what specifically happened around May 1st (a sale event, marketing campaign, or seasonal factor) since it drove nearly double the typical daily order volume.

### Chart 5: Does order quantity relate to order amount?

```python
valid_orders = df[df['Amount'] > 0]

plt.figure(figsize=(10, 6))
plt.scatter(valid_orders['Qty'], valid_orders['Amount'], alpha=0.3, color='purple')
plt.title('Order Quantity vs Order Amount')
plt.xlabel('Quantity')
plt.ylabel('Amount (INR)')
plt.tight_layout()
plt.show()
```

**Insight:** Surprisingly, order Quantity and Amount show almost no clear relationship — orders with a Quantity of 0 or 1 span nearly the same wide price range (roughly ₹200-₹2,500) as orders with a Quantity of 2 or 3. This suggests `Amount` reflects the price of the item(s) purchased rather than scaling directly with quantity — likely because different products (a basic kurta vs. a premium set) vary widely in price regardless of how many units are ordered. The densest cluster sits at Quantity 0-1 with Amount under ₹1,500, confirming most orders are small, single-item, moderately-priced purchases (consistent with Chart 3's findings). A small number of outliers at Quantity 5 and 8 with Amounts over ₹2,000 stand out as bulk or business purchases, distinct from the typical customer pattern.

## EDA Summary

- **Order fulfillment:** ~82% of orders complete successfully (Shipped + Delivered), while 14% are cancelled — a meaningful share worth investigating for root causes.
- **Product concentration:** Just 2 categories (`Set`, `kurta`) drive ~78% of all orders — the business is heavily reliant on a narrow product range.
- **Order value:** Most orders fall between ₹300-₹900, with a right-skewed tail of rare high-value purchases — revenue is driven by volume, not big-ticket sales.
- **Time trend:** A sharp order spike occurred around May 1st (nearly 2x baseline), followed by a settled lower volume through June — worth investigating what drove that spike.
- **Quantity vs. Amount:** No strong relationship between items ordered and order value — price is driven more by product type than quantity purchased.
